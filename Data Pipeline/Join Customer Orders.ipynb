```python
%sql
CREATE OR REPLACE TABLE data_pipeline.customer_orders_silver AS
SELECT 
  c.customer_id,
  c.customer_name,
  c.state,
  c.city,
  o.order_number,
  o.order_datetime,
  o.number_of_line_items,
  o.promo_info
FROM data_pipeline.customer_bronze AS c
JOIN data_pipeline.order_bronze AS o
ON c.customer_id = o.customer_id
```
