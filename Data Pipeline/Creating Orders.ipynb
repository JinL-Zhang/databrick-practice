```python
# Create order_bronze table from sales_orders source
source_table = "databricks_simulated_retail_customer_data.v01.sales_orders"
catalog = "workspace"
schema = "data_pipeline"
target_table = f"{catalog}.{schema}.order_bronze"

# Create schema if it doesn't exist
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")
print(f"✓ Ensured schema exists: {catalog}.{schema}")

# Read from source table
df = spark.table(source_table)

# Create bronze table
df.write.mode("overwrite").saveAsTable(target_table)

print(f"✓ Created table: {target_table}")
print(f"✓ Row count: {df.count()}")
print(f"✓ Columns: {', '.join(df.columns)}")
```
