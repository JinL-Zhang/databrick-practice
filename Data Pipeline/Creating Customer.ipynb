```python
# Get the job/task parameter values
my_catalog = dbutils. widgets.get("catalog")
my_schema = dbutils.widgets.get("schema")
save_schema = dbutils.widgets.get("save_schema")

# Set path to my volume
my_volume_path = f"/Volumes/{my_catalog}/{my_schema}/test_volume/test_directory/customers.csv/"
```

```python
# load the customer.csv file from the volume
customer_df = spark.read.format("csv").option("header", "true").load(my_volume_path)

# load the table
customer_df.write.mode("overwrite").saveAsTable(f"{my_catalog}.{save_schema}.customer_bronze") 
```
