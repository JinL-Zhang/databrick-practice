```python
%sql
CREATE OR REPLACE TABLE data_pipeline.customer_sales_silver AS
SELECT 
  c.customer_id,
  c.customer_name,
  c.loyalty_segment,
  c.units_purchased,
  s.product_name,
  s.order_date,
  s.total_price
FROM data_pipeline.customer_bronze AS c
JOIN data_pipeline.sale_bronze AS s
ON c.customer_id = s.customer_id
```

```python
df = spark.sql('''SELECT * FROM data_pipeline.customer_sales_silver''')

# check for duplicates
duplicate_exists = df.count() > df.dropDuplicates().count()

# Set boolean flag in a job value
dbutils.jobs.taskValues.set(key='has_duplicates', value = duplicate_exists)
```
