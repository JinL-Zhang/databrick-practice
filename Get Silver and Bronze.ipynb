```python
%sql
CREATE OR REFRESH STREAMING LIVE TABLE user_query_silver
COMMENT "Silver layer - cleaned"
AS SELECT 
  UPPER(user_query) as USER_QUERY,
  current_timestamp() as CurrentTimeStamp,
  current_date() as CurrentDate
FROM LIVE.user_query_bronze;

CREATE OR REFRESH STREAMING LIVE TABLE user_query_gold
COMMENT "Gold layer - aggregated by date"
AS SELECT 
  CurrentDate,
  COUNT(*) as total_search
FROM LIVE.user_query_silver
GROUP BY CurrentDate;
```
